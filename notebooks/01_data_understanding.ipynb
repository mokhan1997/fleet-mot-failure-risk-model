## Notebook 01: Data Understanding

This notebook is the first stage of the Fleet MOT Failure Risk Model project.

The purpose of this notebook is to:

- Load the 2024 MOT sample dataset
- Understand the structure of the data
- Review row counts and column counts
- Inspect column names and data types
- Check missing values
- Review MOT test result values
- Identify useful fields for failure risk analysis

At this stage, no cleaning or modelling is being done. The focus is only on understanding the raw sample data.

```python
from pathlib import Path
import pandas as pd
import numpy as np
```

```python
# Define project paths

sample_data_path = Path("../data/sample")
sample_file = sample_data_path / "mot_2024_sample.csv"

print("Sample folder exists:", sample_data_path.exists())
print("Sample file exists:", sample_file.exists())
```

```python
# Load the MOT sample dataset

mot = pd.read_csv(sample_file, low_memory=False)

print("Dataset loaded successfully.")
print("Rows:", mot.shape[0])
print("Columns:", mot.shape[1])
```

```python
# Preview first 5 rows

mot.head()
```

```python
# Preview last 5 rows

mot.tail()
```

```python
# Display all column names

mot.columns.tolist()
```

```python
# Check data types and non-null counts

mot.info()
```

```python
# Basic dataset summary

summary = pd.DataFrame({
    "metric": [
        "rows",
        "columns",
        "duplicate_rows"
    ],
    "value": [
        mot.shape[0],
        mot.shape[1],
        mot.duplicated().sum()
    ]
})

summary
```

```python
# Missing values by column

missing_summary = (
    mot.isna()
    .sum()
    .reset_index()
)

missing_summary.columns = ["column_name", "missing_values"]
missing_summary["missing_percentage"] = round(
    missing_summary["missing_values"] / len(mot) * 100,
    2
)

missing_summary.sort_values(by="missing_values", ascending=False)
```

```python
# Check unique values in likely categorical columns

possible_columns = [
    "test_result",
    "make",
    "model",
    "fuel_type",
    "vehicle_class",
    "colour"
]

for column in possible_columns:
    if column in mot.columns:
        print(f"\n{column.upper()}")
        print("-" * 50)
        print(mot[column].value_counts(dropna=False).head(20))
    else:
        print(f"{column} not found in dataset.")
```

```python
# Summary statistics for numeric columns

mot.describe()
```

```python
# Search for useful column names

keywords = ["result", "date", "odometer", "mileage", "make", "model", "fuel", "class"]

for keyword in keywords:
    matching_columns = [column for column in mot.columns if keyword.lower() in column.lower()]
    print(f"\nColumns containing '{keyword}':")
    print(matching_columns)
```
